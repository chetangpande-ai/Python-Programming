###Encapsulation

```python
#Scope levels in python


class Bank:
    def __init__(self,name,balance):
        self.name = name
        self._account_typoe = "Savings" #protected variable
        self.__balance = balance #private variable


    def get_balance(self):
        return self.__balance

    def add_balance(self,amount):
        self.__balance += amount
```

```python
b1=Bank("SBI",10000)
print(b1.name) #public variable
```

```python
print(b1._account_typoe) #protected variable
```

```python
print(b1.__balance) #private variable, will raise an error
```
