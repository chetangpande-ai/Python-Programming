```python
items = ["api", "ui", "api", "db", "api", "ui"]
print(items.count("api"))  # Output: 3
dict_items = {}

unique_items = set(items)
for item in unique_items:
    count = items.count(item)
    dict_items[item] = count
 
print(dict_items)  # Output: {'db': 1, 'ui': 2, 'api': 3}
```
