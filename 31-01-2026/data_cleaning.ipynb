```python
''' This program cleans and deduplicates a list of strings by removing None values,
empty strings, whitespace, and converting to lowercase, then removes duplicates '''

data = ["  Apple ", None, "banana", "APPLE", "Banana ", "", "orange  "]
type(data)
newdata = []
for item in data:
    if item is not None and item.strip() != "":
        item = item.strip().lower()
        print(item)
        newdata.append(item)

print(newdata)
#remove duplicates from newdata
newdata = list(set(newdata))
print(newdata)
```

```python
#find common elements between two lists

def find_common(a, b):
    set_a = set(a)
    set_b = set(b)
    for x in set_a:
        if x in set_b:
            print(x)

find_common(["apple", "banana", "cherry"], ["banana", "cherry", "date"])
```

```python
a = ["apple", "banana", "cherry", "date", "fig", "grape", "kiwi","lemon","mango"]
b = ["banana", "cherry", "date", "orange", "papaya", "quince","raspberry","strawberry","tangerine"]
common = set(a) & set(b)
print(common)  # Output: {'banana', 'cherry'}
```
