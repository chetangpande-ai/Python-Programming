### Inheritance

```python
class Speciality():
    def neuro_surgegy(self):
        print("Speciality- neuro surgon")

    def heart_surgery(self):
        print("speciality-heart surgery")


class GeneralHospital(Speciality):
    def opd(self):
        print("general hospital-opd")

```

```python
g1  = GeneralHospital()
g1.heart_surgery()
g1.neuro_surgegy()
g1.opd()
```
