### Super

```python
class Employee():
    def __init__(self,name):
        self.name=name
        print("student name is -",self.name)


class Manager(Employee):
    def __init__(self,name,teamSize):
        super().__init__(name)
        self.teamSize=teamSize
        print("teamSize - ",self.teamSize)


```

```python
e1 = Manager("Amit", 5)
```
