# OOPs(Classes & Objects)

```python
#Car
# -customer name
# -car model
# -car number
# -car color
class Car():
    pass

    def __init__(self,car_name, car_model, car_number, car_color):
        self.a=2
        self.car_name=car_name
        self.car_model=car_model
        self.car_number=car_number
        self.car_color=car_color
        print("This is constructor")

    def print_car_name(self):
        print(self.car_name)

    def print_car_details(self):
        print("car name-",self.car_name,", model-", self.car_model," ,number-", self.car_number, ", color-",self.car_color)
```

```python
car1=Car("BMW","X5","DL-12-AB-1234","Black")
car1.print_car_name()
```

```python
print(id(car1))
print(type(car1))
```
