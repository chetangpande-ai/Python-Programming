```python
class Teacher:
    def __init__(self,name:str,age:int):
        self.name=name
        self.age=age

    def printTeacherInfo(self):
        print("name-",self.name,",age-",self.age)


t1=Teacher("Amit",12)
print(t1)
```

```python
t1.printTeacherInfo
```
